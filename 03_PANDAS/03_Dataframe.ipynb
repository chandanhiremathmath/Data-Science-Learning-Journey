Because real-world datasets are usually stored in rows and columns, DataFrames are the primary structure used for data analysis, data cleaning, feature engineering, and machine learning.

# 3.1 Creating a DataFrame from a Dictionary

## 📖 Definition

A DataFrame can be created from a **Python dictionary**, where:

- Each **key** becomes a column name.
- Each **value** becomes the data of that column.
- Every value should generally be a list (or another iterable) of the same length.

This is the **most common and beginner-friendly** way to create a DataFrame.

---

## 🤔 Why do we need it?

While learning Pandas, we often don't have a CSV file ready.

Instead of downloading a dataset every time, we can quickly create our own dataset using a dictionary.

It is also useful for:

- Testing new code
- Creating sample datasets
- Demonstrating examples
- Unit testing
- Interview coding questions

---

## 🧠 Think Like a Data Scientist

Imagine your manager asks you to quickly test a new data cleaning function.

Instead of waiting for the production dataset, you create a small sample dataset using a dictionary and verify your logic in a few seconds.

This is why Data Scientists frequently use dictionaries while experimenting.

---

## 📌 Syntax

```python
pd.DataFrame(data)
```

or

```python
pd.DataFrame(dictionary_name)
```

where:

- `pd` → Alias for the Pandas library.
- `DataFrame` → Pandas class used to create a table.
- `data` → Dictionary containing the dataset.

````python
import pandas as pd
````

# 3.2 Creating a DataFrame with a Custom Index

## 📖 Definition

By default, Pandas assigns an integer index starting from `0`.

However, we can provide our own meaningful row labels while creating a DataFrame. These labels are called **custom indexes**.

A custom index makes the dataset easier to understand and access, especially when the rows already have unique identifiers.

---

## 🤔 Why do we need a Custom Index?

In real-world datasets, rows often have unique identifiers instead of simple numbers.

Examples:

- Student ID
- Employee ID
- Product ID
- Order ID
- Customer ID

Instead of:

```
0
1
2
3
```

we can use:

```
S101
S102
S103
S104
```

This makes the data more meaningful and easier to work with.

---

## 🧠 Think Like a Data Scientist

Imagine a company has 50,000 employees.

Finding an employee using row number `12546` is not practical.

Instead, employees are identified using IDs like:

```
EMP1001
EMP1002
EMP1003
```

These IDs become the DataFrame index.

---

## 📌 Syntax

```python
pd.DataFrame(data, index=index_labels)
```

where:

- `data` → Dictionary or other data source.
- `index` → List containing custom row labels.

````python
student_data = {
    "Name": ["Rahul", "Priya", "Amit", "Sneha"],
    "Age": [21, 20, 22, 21],
    "Department": ["CSE", "ISE", "ECE", "CSE"],
    "Marks": [92, 87, 78, 95]
}

student_data
````

## Creating a DataFrame with Default Index

````python
import pandas as pd
df_default = pd.DataFrame(student_data)

df_default
````

## Creating a DataFrame with Custom Index

````python
student_index = ["S101", "S102", "S103", "S104"]

df = pd.DataFrame(student_data, index=student_index)

df
````

## 🔍 Output Explanation

Let's understand what changed.

### Before

```

1
2
3
```

These were automatically generated by Pandas.

### After

```

S102
S103
S104
```

These are the custom indexes we supplied.

Notice that:

- The data has not changed.
- Only the row labels have changed.
- Each row now has a meaningful identifier.

# 3.3 DataFrame Attribute - `shape`

## 📖 Definition

The `shape` attribute returns the **number of rows and columns** present in a DataFrame.

It returns the result as a **tuple** in the form:

```python
(rows, columns)
```

Unlike methods such as `head()` or `describe()`, `shape` is an **attribute**, so it is accessed **without parentheses**.

---

## 🤔 Why do we use `shape`?

Whenever we receive a new dataset, the first question is:

- How many records are present?
- How many features (columns) are available?

Instead of counting them manually, we simply use:

```python
df.shape
```

This instantly tells us the size of the dataset.

---

## 🧠 Think Like a Data Scientist

Imagine your manager sends you a customer dataset.

Before doing any analysis, you need to know:

- Is it a small dataset?
- Is it a large dataset?
- How many features are available?

The first command you will most likely run is:

```python
df.shape
```

This gives you a quick overview of the dataset.

````python
import pandas as pd

student_data = {
    "Name": ["Rahul", "Priya", "Amit", "Sneha"],
    "Age": [21, 20, 22, 21],
    "Department": ["CSE", "ISE", "ECE", "CSE"],
    "Marks": [92, 87, 78, 95]
}

df = pd.DataFrame(student_data)

df
````

## 📌 Syntax

```python
df.shape
```

Since `shape` is an **attribute**, we do **not** write parentheses.

❌ Incorrect

```python
df.shape()
```

✅ Correct

```python
df.shape
```

````python
df.shape
````

## 📤 Expected Output

```python
(4, 4)
```

---

## 🔍 Output Explanation

The output is a tuple:

```
(rows, columns)
```

For our DataFrame:

- Rows = **4**
- Columns = **4**

Therefore:

```python
(4, 4)
```

### 🧩 Visualization

```
            Name    Age   Department   Marks
---------------------------------------------

Row 2       Priya    20      ISE         87
Row 3       Amit     22      ECE         78
Row 4       Sneha    21      CSE         95
---------------------------------------------
            ↑
         4 Rows

Columns →

Name | Age | Department | Marks

= 4 Columns

Shape = (4, 4)
```

````python
rows, columns = df.shape

print("Number of Rows    :", rows)
print("Number of Columns :", columns)
````

## 🔍 Explanation

Since `shape` returns a tuple, we can unpack it into two variables.

```python
rows, columns = df.shape
```

Now:

- `rows` stores the number of rows.
- `columns` stores the number of columns.

This is commonly used in machine learning pipelines and data validation scripts.

# 3.4 DataFrame Attribute - `size`

## 📖 Definition

The `size` attribute returns the **total number of elements** present in a DataFrame.

It is calculated as:

```python
Number of Rows × Number of Columns
```

Unlike `shape`, which returns two values, `size` returns a **single integer**.

`size` is an **attribute**, so it is accessed **without parentheses**.

---

## 🤔 Why do we use `size`?

Sometimes we want to know the total number of values stored in a dataset.

Instead of manually multiplying the number of rows and columns, Pandas provides the `size` attribute.

This is useful for:

- Quickly understanding dataset size
- Data validation
- Comparing datasets before and after preprocessing

---

## 🧠 Think Like a Data Scientist

Suppose your team receives two datasets.

Dataset A

```
500 rows
8 columns
```

Dataset B

```
1000 rows
12 columns
```

To estimate how much data each contains, one quick metric is the **total number of elements**, which you can obtain using:

```python
df.size
```

````python
import pandas as pd

student_data = {
    "Name": ["Rahul", "Priya", "Amit", "Sneha"],
    "Age": [21, 20, 22, 21],
    "Department": ["CSE", "ISE", "ECE", "CSE"],
    "Marks": [92, 87, 78, 95]
}

df = pd.DataFrame(student_data)

df
````

## 📌 Syntax

```python
df.size
```

Since `size` is an **attribute**, do **not** use parentheses.

✅ Correct

```python
df.size
```

❌ Incorrect

```python
df.size()
```

````python
df.size
````

## 📤 Expected Output

```python
16
```

---

## 🔍 Output Explanation

Our DataFrame contains:

- Rows = **4**
- Columns = **4**

Therefore,

```
Total Elements

=

Rows × Columns

=

4 × 4

=

16
```

### 🧩 Visualization

```
            Name   Age   Department   Marks
--------------------------------------------
Rahul      1      2         3          4

Priya      5      6         7          8

Amit       9     10        11         12

Sneha     13     14        15         16
--------------------------------------------

Total Elements = 16
```

````python
print("Shape :", df.shape)
print("Size  :", df.size)
````

## 🔍 Understanding the Difference

Output:

```python
Shape : (4, 4)
Size  : 16
```

Explanation:

- `shape` tells us the **structure** of the DataFrame (rows and columns).
- `size` tells us the **total number of values** stored in the DataFrame.

Formula:

```python
df.size == df.shape[0] * df.shape[1]
```

````python
rows, columns = df.shape

print("Rows :", rows)
print("Columns :", columns)

print("Calculated Size :", rows * columns)
print("Actual Size     :", df.size)
````
