# 03. DataFrame

## 📖 Introduction

A **DataFrame** is the most important data structure in Pandas.

It represents data in a **two-dimensional tabular format**, similar to an Excel spreadsheet or a SQL table.

A DataFrame consists of **rows and columns**, where:

- Each **column** is a Pandas Series.
- All columns together form a DataFrame.

Because real-world datasets are usually stored in rows and columns, DataFrames are the primary structure used for data analysis, data cleaning, feature engineering, and machine learning.


# 🎯 Problem Statement

Suppose a school stores student information.

For each student, the following details are recorded:

- Student Name
- Age
- Department
- Marks

If we use a Series, we can store only one column at a time.

However, in real applications, we need to store multiple related columns together.

This problem is solved using a **DataFrame**.

# 🤔 What is a DataFrame?

A **DataFrame** is a two-dimensional labeled data structure provided by Pandas.

It stores data in the form of **rows and columns**, making it similar to an Excel sheet or a database table.

Each column in a DataFrame is actually a Pandas Series.

### Visual Representation

| Name | Age | Department | Marks |
|------|----:|------------|------:|
| Rahul | 21 | CSE | 92 |
| Priya | 20 | ISE | 87 |
| Amit | 22 | ECE | 78 |
| Sneha | 21 | CSE | 95 |

Here:

- Rows represent individual records.
- Columns represent different features (attributes).

# ⚖️ Difference Between Series and DataFrame

| Series | DataFrame |
|--------|-----------|
| One-dimensional | Two-dimensional |
| Single column | Multiple columns |
| Represents one feature | Represents an entire dataset |
| Can have one data type | Different columns can have different data types |
| Suitable for simple data | Suitable for real-world datasets |
